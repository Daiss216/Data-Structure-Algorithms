### Dynamic Array (basic)
includes features ~ creating list, len, append, print, indexing, pop, clear, find, insert, delete, remove, sort/min/max/sum extend, slicing, Merge

```python
#foreign func lib for py provides C compatible data types
import ctypes
```

```python
class MyList():
    def __init__(self):
        self.size= 1    #size
        self.n= 0       #no. of elements
## create Ctype array with size= self.size
        self.A= self.__make_array(self.size)

##LENGTH
    def __len__(self):
        return self.n

##PRINT
    def __str__(self):
        result= ''
        
        for i in range(self.n):
            result= result + str(self.A[i]) + ','
        return '[' + result[:-1] + ']'

    def __getitem__(self, index):
        if 0 <= index <= self.n:
            return self.A[index]
        else:
            return 'IndexError- Index out of range'
            
##APPEND        
    def append(self, item):
        if self.n == self.size:
            self.__resize(self.size*2)  #resize
        #else append
        self.A[self.n] = item
        self.n = self.n + 1

##DELETING LAST  ELEMENT
    def pop(self):
        if self.n == 0:
            return 'IndexError - List is empty'
        print(self.A[self.n - 1])
        self.n= self.n -1

##CLEARING LIST
    def clear(self):
        self.size= 1
        self.n= 0

##FINDING INDEX
    def find(self, item):
        for i in range(self.n):
            if self.A[i] == item:
                return i
            #return 'ValueError - Item not found'
##INSERTION
    def insert(self,pos,item):
        if 0 <= pos < self.n:  #[0 - no. of elements]
            if self.n == self.size:      #resize
                self.__resize(self.size*2)

            for i in range(self.n, pos, -1):  # -ve looping
                self.A[i] = self.A[i-1]

            self.A[pos] = item
            self.n = self.n + 1

        else:
            return 'IndexError'

##REMOVE ITEM -> delete based on value
    def remove(self, item):
        pos = self.find(item)

        if pos != 'ValueError - Item not found':
            #delete
            self.__delitem__(pos)
        else:
            return pos

##DELETION -> delete besed on index
    def __delitem__(self,pos):
        if 0<= pos <self.n:
            for i in range(pos, self.n - 1):
                self.A[i] = self.A[i+1]
            
            self.n = self.n - 1

#REDEFINE SIZE
    def __resize(self, new_capacity):
        #resize
        B= self.__make_array(new_capacity)
        self.size= new_capacity

        #copy
        for i in range(self.n):
            B[i]= self.A[i]
  
        #reassign
        self.A = B

##CREATING ARRAY
    def __make_array(self, capacity):
        # create Ctype array(static, referential) with size capacity
        return (capacity*ctypes.py_object)()        ## C code in py
```
